```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('../data/mental_health_dataset.csv')
print(data.head())
```

```python
print(data.info())
```

```python
print(data.describe())
```

```python
corr_matrix = data.corr(numeric_only=True)
```

```python
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
```

```python
plt.hist(data['age'], bins=30, edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

plt.hist(data['sleep_hours'], bins=30, edgecolor='black')
plt.title('Sleep Distribution')
plt.xlabel('Sleep')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.hist(data['stress_level'], bins=30, edgecolor='black')
plt.title('Stress Level Distribution')
plt.xlabel('Stress Level')
plt.ylabel('Frequency')
plt.show()

plt.hist(data['physical_activity_days'], bins=30, edgecolor='black')
plt.title('Physical Activity Days Distribution')
plt.xlabel('Physical Activity Days')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.hist(data['depression_score'], bins=30, edgecolor='black')
plt.title('Depression Score Distribution')
plt.xlabel('Depression Score')
plt.ylabel('Frequency')
plt.show()

plt.hist(data['anxiety_score'], bins=30, edgecolor='black')
plt.title('Anxiety Score Distribution')
plt.xlabel('Anxiety Score')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.hist(data['social_support_score'], bins=30, edgecolor='black')
plt.title('Social Support Score Distribution')
plt.xlabel('Social Support Score')
plt.ylabel('Frequency')
plt.show()

plt.hist(data['productivity_score'], bins=30, edgecolor='black')
plt.title('Productivity Score Distribution')
plt.xlabel('Productivity Score')
plt.ylabel('Frequency')
plt.show()
```
