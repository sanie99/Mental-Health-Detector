```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score 
```

```python
data = pd.read_csv('../data/processed_mental_health_dataset.csv')
print(data.head())
```

```python
X, y = data.drop('mental_health_risk', axis=1), data['mental_health_risk']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=40, stratify=y)
```

```python
logistic_model = LogisticRegression(max_iter=1000)
logistic_model.fit(X_train, y_train)
logistic_y_pred = logistic_model.predict(X_test)
accuracy = accuracy_score(y_test, logistic_y_pred)
print(f"Accuracy of the Logistic Regression model: {accuracy}")
```

```python
from sklearn.tree import DecisionTreeClassifier
decision_tree_model = DecisionTreeClassifier(random_state=40)
decision_tree_model.fit(X_train, y_train)
decision_tree_y_pred = decision_tree_model.predict(X_test)
decision_tree_accuracy = accuracy_score(y_test, decision_tree_y_pred)
print(f"Accuracy of the Decision Tree model: {decision_tree_accuracy}")
```

```python
from sklearn.ensemble import RandomForestClassifier
random_forest_model = RandomForestClassifier(n_estimators=200 ,random_state=40)
random_forest_model.fit(X_train, y_train)
random_forest_y_pred = random_forest_model.predict(X_test)
random_forest_accuracy = accuracy_score(y_test, random_forest_y_pred)
print(f"Accuracy of the Random Forest model: {random_forest_accuracy}")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print("Logistic Regression Classification Report:")
print(classification_report(y_test, logistic_y_pred))

print("Decision Tree Classification Report:")
print(classification_report(y_test, decision_tree_y_pred))

print("Random Forest Classification Report:")
print(classification_report(y_test, random_forest_y_pred))
```

```python
print("Logistic Regression Confusion Matrix:")
print(confusion_matrix(y_test, logistic_y_pred))

print("Decision Tree Confusion Matrix:")
print(confusion_matrix(y_test, decision_tree_y_pred))

print("Random Forest Confusion Matrix:")
print(confusion_matrix(y_test, random_forest_y_pred))
```

```python
from sklearn.model_selection import cross_val_score
```

```python
logistic_cv_scores = cross_val_score(logistic_model, X_train, y_train, cv=5)
decision_tree_cv_scores = cross_val_score(decision_tree_model, X_train, y_train, cv=5)
random_forest_cv_scores = cross_val_score(random_forest_model, X_train, y_train, cv=5)
```

```python
cross_val_results = {
    "Logistic Regression": logistic_cv_scores,
    "Decision Tree": decision_tree_cv_scores,
    "Random Forest": random_forest_cv_scores
}

print("Cross-Validation Results:")
for model_name, scores in cross_val_results.items():
    print(f"{model_name}: Mean Accuracy = {scores.mean():.4f}")
```

```python
import joblib
joblib.dump(logistic_model, '../models/logistic_model.pkl')
joblib.dump(decision_tree_model, '../models/decision_tree_model.pkl')
joblib.dump(random_forest_model, '../models/random_forest_model.pkl')
print("Models saved successfully.")
```
