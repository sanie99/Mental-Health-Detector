```python
import pandas as pd
```

```python
data = pd.read_csv('../data/mental_health_dataset.csv')
print(data.head())
```

```python
data['gender'] = data["gender"].map({"Male": 0, "Female": 1, "Non-binary": 2, "Prefer not to say": 3})
```

```python
print(data['work_environment'].unique())
```

```python
data['employment_status'] = data["employment_status"].map({"Employed": 0, "Unemployed": 1, "Student": 2, "Self-employed": 3})
print(data.head())
```

```python
data['work_environment'] = data["work_environment"].map({"Remote": 0, "On-site": 1, "Hybrid": 2})
print(data.head())
```

```python
data['mental_health_history'] = data["mental_health_history"].map({"Yes": 1, "No": 0})
print(data.head())
```

```python
data['seeks_treatment'] = data["seeks_treatment"].map({"Yes": 1, "No": 0})
print(data.head())
```

```python
data['mental_health_risk']  = data["mental_health_risk"].map({"Low": 0, "Medium": 1, "High": 2})
print(data.head())
```

```python
import seaborn as sns
sns.heatmap(data.corr(), annot=True, cmap='coolwarm')
```

```python
for column in data.columns:
    if data[column].corr(data['mental_health_risk']) < 0.1:
        data.drop(column, axis=1, inplace=True)
print(data.head())
```

```python
data.to_csv('../data/processed_mental_health_dataset.csv', index=False)
print("Data preprocessing completed and saved to processed_mental_health_dataset.csv")
```
